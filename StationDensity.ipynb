```python
def stations_by_country(country_code):
    return stations.region(country_code).count()
```

```python
def build_station_density(country_dict):
    country_df = pd.DataFrame.from_dict(
        country_dict, orient='index', columns=['Area (km²)']
    )
    country_df['Station Count'] = [stations_by_country(code) for code in tqdm(country_dict.keys())]
    country_df['Density (stations per 10,000 km²)'] = (
        country_df['Station Count'] / country_df['Area (km²)'] * 100_000
    )
    # enforce column order
    country_df = country_df[['Area (km²)', 'Station Count', 'Density (stations per 10,000 km²)']]
    # sort by density, highest first
    country_df = country_df.sort_values(by='Density (stations per 10,000 km²)', ascending=False)
    return country_df
```
